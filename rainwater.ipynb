```python
def rainwater(lst):
    if not lst:
        return 0

    n = len(lst)
    left_max = [0] * n
    right_max = [0] * n

    left_max[0] = lst[0]
    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], lst[i])

    right_max[n - 1] = lst[n - 1]
    for i in range(n - 2, -1, -1):
        right_max[i] = max(right_max[i + 1], lst[i])

    water_trapped = 0
    for i in range(n):
        water_trapped += min(left_max[i], right_max[i]) - lst[i]

    return water_trapped
```

```python
rainwater([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1])
```

```python
rainwater([4, 2, 0, 3, 2, 5])
```

```python
rainwater([2, 0, 2])
```

```python
rainwater([3, 0, 2, 0, 4])
```
